### Step 1. Import the necessary libraries

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called 

```python
url = 'https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv'
chipo = pd.read_csv(url, sep='\t')
chipo['item_price'] = chipo['item_price'].str.replace('$', '', regex=False).astype(float)
chipo.head()
```

### Step 4. 

```python
# Step 4. Select the 'item_name' column by label
chipo['item_name'].head()
```

### Step 5. 

```python
# Step 5. Select several columns by label
chipo[['order_id', 'item_name', 'item_price']].head()
```

### Step 6. 

```python
# Step 6. Select rows 10 to 14 by position with iloc
chipo.iloc[10:15]
```

### Step 7. 

```python
# Step 7. Select rows by index label with loc
chipo.loc[100:104, ['item_name', 'quantity']]
```

### Step 8. 

```python
# Step 8. Select a single cell by position and by label
print(chipo.iat[0, 2])
print(chipo.at[0, 'item_name'])
```

### Step 9. 

```python
# Step 9. Boolean indexing: lines with quantity greater than 2
chipo[chipo['quantity'] > 2]
```

### Step 10. 

```python
# Step 10. Combine conditions: Chicken Bowl lines costing more than $10
chipo[(chipo['item_name'] == 'Chicken Bowl') & (chipo['item_price'] > 10)]
```

### Step 11. 

```python
# Step 11. Set order_id as the index
by_order = chipo.set_index('order_id')
by_order.head()
```

### Step 12. 

```python
# Step 12. Select all lines of order 1 using the new index
by_order.loc[1]
```

### Step 13. 

```python
# Step 13. Create a MultiIndex of order_id and item_name
multi = chipo.set_index(['order_id', 'item_name']).sort_index()
multi.head()
```

### Step 14. 

```python
# Step 14. Select from the MultiIndex: order 4, item 'Steak Burrito'
multi.loc[(4, 'Steak Burrito')]
```

### Step 15. 

```python
# Step 15. Use query() to filter with an expression
chipo.query('quantity >= 3 and item_price < 15')
```

### Step 16. 

```python
# Step 16. Reset the index back to a default RangeIndex
by_order.reset_index().head()
```
