### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Titanic_Desaster/train.csv)

### Step 3. Assign it to a variable titanic 

```python
# The online copy returns 404, so we use the train.csv shipped with the exercise
titanic = pd.read_csv('train.csv')
titanic.head()
```

### Step 4. Set PassengerId as the index 

```python
titanic = titanic.set_index('PassengerId')
titanic.head()
```

### Step 5. Create a pie chart presenting the male/female proportion

```python
counts = titanic['Sex'].value_counts()
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=90, explode=(0.1, 0), shadow=True)
plt.axis('equal')
plt.title('Sex proportion')
plt.show()
```

### Step 6. Create a scatterplot with the Fare payed and the Age, differ the plot color by gender

```python
sns.scatterplot(x='Fare', y='Age', hue='Sex', data=titanic)
plt.title('Fare x Age')
plt.show()
```

### Step 7. How many people survived?

```python
titanic['Survived'].sum()
```

### Step 8. Create a histogram with the Fare payed

```python
plt.hist(titanic['Fare'], bins=np.arange(0, 600, 10))
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.title('Fare Paid Histogram')
plt.show()
```

### BONUS: Create your own question and answer it.

```python
# BONUS: survival rate by passenger class and sex
sns.barplot(x='Pclass', y='Survived', hue='Sex', data=titanic)
plt.ylabel('Survival rate')
plt.title('Survival rate by class and sex')
plt.show()
```
